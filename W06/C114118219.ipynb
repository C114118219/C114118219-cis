```python
import numpy as np
import pandas as pd
random_seed = 42
rng = np.random.default_rng(random_seed)

student_count = 60
n = student_count

# 1. 學號 1~60
student_id = np.arange(1, n + 1)

# 2. 科系隨機抽樣
departments_list = ["資管系", "企管系", "財金系"]
department = rng.choice(departments_list, size=n)

# 3. 睡眠時數：平均 6.7，標準差 1.1，四捨五入至小數一位
sleep_hours = np.round(rng.normal(loc=6.7, scale=1.1, size=n), 1)

# 4. 螢幕時數：平均 5.0，標準差 1.5，限制在 1~10 小時
screen_time_hours = np.round(
    np.clip(rng.normal(loc=5.0, scale=1.5, size=n), 1.0, 10.0), 1
)

# 5. 學習時數：平均 2.5，標準差 1.0，限制在 0.5~6 小時
study_hours = np.round(
    np.clip(rng.normal(loc=2.5, scale=1.0, size=n), 0.5, 6.0), 1
)

# 6. 出席率：平均 88，標準差 8，限制在 60~100，取整數
attendance_rate = np.round(
    np.clip(rng.normal(loc=88, scale=8, size=n), 60, 100)
).astype(int)

# 組合成 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 查看前 5 筆
df.head()
# 產生平均為 0，標準差為 5 的隨機雜訊
noise = rng.normal(loc=0, scale=5, size=n)

# 向量化公式計算
raw_score = (
    50 
    + (3 * df["study_hours"]) 
    + (2 * df["sleep_hours"]) 
    - (1.2 * df["screen_time_hours"]) 
    + (0.15 * df["attendance_rate"]) 
    + noise
)

# 邊界限制 0~100 並轉為整數
df["quiz_score"] = np.round(np.clip(raw_score, 0, 100)).astype(int)

# 顯示關鍵欄位
df[["student_id", "study_hours", "sleep_hours", "screen_time_hours", "attendance_rate", "quiz_score"]].head(60)
```
